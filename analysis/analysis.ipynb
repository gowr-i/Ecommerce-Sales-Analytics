## 5. Sales & Profit Analysis

This section analyzes the overall sales and profitability of the e-commerce business using total sales, total profit, average order value, average profit per order, and profit margin.

```python
total_sales = df['Total_Sales'].sum()
total_profit = df['Profit'].sum()
average_sales = df['Total_Sales'].mean()
average_profit = df['Profit'].mean()
profit_margin = (total_profit / total_sales) * 100

print(f"Total Sales: {total_sales:,.2f}")
print(f"Total Profit: {total_profit:,.2f}")
print(f"Average Order Value: {average_sales:,.2f}")
print(f"Average Profit per Order: {average_profit:,.2f}")
print(f"Profit Margin: {profit_margin:.2f}%")
```

### Key Finding

The analysis shows a profit margin of 32.79%, indicating that the business generates approximately 32.79 units of profit for every 100 units of sales. Total sales and total profit provide an overall view of the business's financial performance.

```python
metrics = [total_sales, total_profit]
labels = ['Total Sales', 'Total Profit']

plt.figure(figsize=(7, 5))
plt.bar(labels, metrics)
plt.title('Total Sales vs Total Profit')
plt.ylabel('Amount')
plt.tight_layout()

plt.savefig("../visualizations/sales_vs_profit.png", dpi=300)
plt.show()
```
